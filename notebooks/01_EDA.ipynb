```python
import os
import matplotlib.pyplot as plt
import random
from PIL import Image
```

```python
DATASET_PATH = "../data/food-101/images"
```

```python
CLASSES = [
    "pizza",
    "hamburger",
    "ceviche",
    "tacos",
    "steak",
    "ramen",
    "ice_cream",
    "spaghetti_bolognese",
    "fried_rice",
    "chicken_wings"
]

print("Clases seleccionadas:", CLASSES)
print("Total:", len(CLASSES))
```

```python
clases_validas = []

for clase in CLASSES:
    ruta = os.path.join(DATASET_PATH, clase)
    
    if os.path.exists(ruta):
        clases_validas.append(clase)
    else:
        print(f"⚠️ Clase no encontrada: {clase}")

CLASSES = clases_validas

print("Clases finales:", CLASSES)
```

```python
conteo = {}

for clase in CLASSES:
    ruta = os.path.join(DATASET_PATH, clase)
    conteo[clase] = len(os.listdir(ruta))

print("Conteo por clase:")
print(conteo)
```

```python
clase_random = random.choice(CLASSES)
ruta_clase = os.path.join(DATASET_PATH, clase_random)

imagen_random = random.choice(os.listdir(ruta_clase))
ruta_imagen = os.path.join(ruta_clase, imagen_random)

img = Image.open(ruta_imagen)

plt.imshow(img)
plt.title(clase_random)
plt.axis("off")
plt.show()
```

```python
plt.figure(figsize=(10, 10))

for i in range(9):
    clase_random = random.choice(CLASSES)
    ruta_clase = os.path.join(DATASET_PATH, clase_random)

    imagen_random = random.choice(os.listdir(ruta_clase))
    ruta_imagen = os.path.join(ruta_clase, imagen_random)

    img = Image.open(ruta_imagen)

    plt.subplot(3, 3, i+1)
    plt.imshow(img)
    plt.title(clase_random)
    plt.axis("off")

plt.show()
```

```python
plt.figure(figsize=(8, 8))

for i in range(4):
    clase_random = random.choice(CLASSES)
    ruta_clase = os.path.join(DATASET_PATH, clase_random)

    imagen_random = random.choice(os.listdir(ruta_clase))
    ruta_imagen = os.path.join(ruta_clase, imagen_random)

    img = Image.open(ruta_imagen).resize((128, 128))

    plt.subplot(2, 2, i+1)
    plt.imshow(img)
    plt.title(clase_random)
    plt.axis("off")

plt.show()
```
