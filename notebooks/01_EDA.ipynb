```python
import os
import random
import matplotlib.pyplot as plt
from PIL import Image
```

```python
DATASET_PATH = "../data/food-101/images"
```

```python
CLASSES = sorted([
    d for d in os.listdir(DATASET_PATH)
    if os.path.isdir(os.path.join(DATASET_PATH, d))
])

print("Total clases:", len(CLASSES))
print(CLASSES[:10])
```

```python
plt.figure(figsize=(10,10))

for i in range(9):
    clase = random.choice(CLASSES)
    ruta_clase = os.path.join(DATASET_PATH, clase)
    
    img_name = random.choice(os.listdir(ruta_clase))
    img_path = os.path.join(ruta_clase, img_name)

    img = Image.open(img_path)

    plt.subplot(3,3,i+1)
    plt.imshow(img)
    plt.title(clase)
    plt.axis("off")

plt.tight_layout()
plt.show()
```
